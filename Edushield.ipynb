## **Machine Learning Model**

```python
# Separate Features and Target Variable

# X contains all input features
X = df.drop(["Student_ID", "Academic_Risk"], axis=1)

# y contains the target variable
y = df["Academic_Risk"]

# Display shape
print("Features Shape :", X.shape)
print("Target Shape :", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X = pd.get_dummies(X, drop_first=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
```

```python
print(X_train.columns.tolist())
```

```python
print(X_train.select_dtypes(include=["object"]).columns.tolist())
```

```python
# Scale Numerical Features
from sklearn.preprocessing import StandardScaler

# Create scaler object
scaler = StandardScaler()

# Fit and transform training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform testing data
X_test_scaled = scaler.transform(X_test)
```

```python
# Import Evaluation Metrics

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    roc_auc_score
)
```

```python
# Logistic Regression

from sklearn.linear_model import LogisticRegression

# Create model
logistic_model = LogisticRegression(random_state=42)

# Train model
logistic_model.fit(X_train_scaled, y_train)

# Make predictions
y_pred_lr = logistic_model.predict(X_test_scaled)

# Calculate accuracy
accuracy_lr = accuracy_score(y_test, y_pred_lr)

print("Logistic Regression Accuracy :", round(accuracy_lr*100,2),"%")
```

```python
# Logistic Regression Evaluation

print(classification_report(y_test, y_pred_lr))

cm = confusion_matrix(y_test, y_pred_lr)

ConfusionMatrixDisplay(cm).plot(cmap="Blues")

plt.title("Logistic Regression Confusion Matrix")

plt.show()
```

```python
#  Decision Tree Classifier

from sklearn.tree import DecisionTreeClassifier

decision_tree = DecisionTreeClassifier(random_state=42)

decision_tree.fit(X_train, y_train)

y_pred_dt = decision_tree.predict(X_test)

accuracy_dt = accuracy_score(y_test, y_pred_dt)

print("Decision Tree Accuracy :", round(accuracy_dt*100,2),"%")
```

```python
# Decision Tree Evaluation

print(classification_report(y_test, y_pred_dt))

cm = confusion_matrix(y_test, y_pred_dt)

ConfusionMatrixDisplay(cm).plot(cmap="Greens")

plt.title("Decision Tree Confusion Matrix")

plt.show()
```

```python
# Random Forest Classifier

from sklearn.ensemble import RandomForestClassifier

random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

random_forest.fit(X_train, y_train)

y_pred_rf = random_forest.predict(X_test)

accuracy_rf = accuracy_score(y_test, y_pred_rf)

print("Random Forest Accuracy :", round(accuracy_rf*100,2),"%")
```

```python
#  Random Forest Evaluation

print(classification_report(y_test, y_pred_rf))

cm = confusion_matrix(y_test, y_pred_rf)

ConfusionMatrixDisplay(cm).plot(cmap="Oranges")

plt.title("Random Forest Confusion Matrix")

plt.show()
```

```python
# Compare Model Accuracy

results = pd.DataFrame({

    "Model":[
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "Accuracy":[
        accuracy_lr,
        accuracy_dt,
        accuracy_rf
    ]

})

results
```

```python
# Accuracy Comparison Graph

plt.figure(figsize=(8,5))

sns.barplot(
    data=results,
    x="Model",
    y="Accuracy",
    palette="viridis"
)

plt.ylim(0,1)

plt.title("Model Accuracy Comparison")

for index,value in enumerate(results["Accuracy"]):

    plt.text(
        index,
        value+0.01,
        f"{value:.2f}",
        ha="center",
        fontsize=10
    )
plt.savefig("model_accuracy.png", dpi=300)
plt.show()
```

```python
# Function to Evaluate Machine Learning Models

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

def evaluate_model(model_name, y_true, y_pred):

    print("="*60)
    print(f"{model_name} Performance")
    print("="*60)

    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    roc_auc = roc_auc_score(y_true, y_pred)

    print("Accuracy :", round(accuracy,4))
    print("Precision:", round(precision,4))
    print("Recall   :", round(recall,4))
    print("F1 Score :", round(f1,4))
    print("ROC-AUC  :", round(roc_auc,4))

    print("\nClassification Report\n")
    print(classification_report(y_true, y_pred))

    cm = confusion_matrix(y_true, y_pred)

    ConfusionMatrixDisplay(cm).plot(cmap="Blues")
    plt.title(f"{model_name} Confusion Matrix")
    plt.show()

    return accuracy, precision, recall, f1, roc_auc
```

```python
# evaluate logistic regression
lr_results = evaluate_model(
    "Logistic Regression",
    y_test,
    y_pred_lr
)
```

```python
# Evaluate decision tree
dt_results = evaluate_model(
    "Decision Tree",
    y_test,
    y_pred_dt
)
```

```python
# Evaluate random forest
rf_results = evaluate_model(
    "Random Forest",
    y_test,
    y_pred_rf
)
```

```python
# Create comparision table
comparison = pd.DataFrame({

    "Model":[
        "Logistic Regression",
        "Decision Tree",
        "Random Forest"
    ],

    "Accuracy":[
        lr_results[0],
        dt_results[0],
        rf_results[0]
    ],

    "Precision":[
        lr_results[1],
        dt_results[1],
        rf_results[1]
    ],

    "Recall":[
        lr_results[2],
        dt_results[2],
        rf_results[2]
    ],

    "F1 Score":[
        lr_results[3],
        dt_results[3],
        rf_results[3]
    ],

    "ROC-AUC":[
        lr_results[4],
        dt_results[4],
        rf_results[4]
    ]

})

comparison
```

```python
from xgboost import XGBClassifier

# Calculate scale_pos_weight to address the 85/15 target imbalance
scale_pos_weight_value = (y_train == 0).sum() / (y_train == 1).sum()

# Create XGBoost model with corrected hyper-parameters
xgb_model = XGBClassifier(
    random_state=42,
    n_estimators=100,
    max_depth=4,                             # Reduced to prevent overfitting
    learning_rate=0.05,                      # Smoother convergence
    subsample=0.8,                           # Prevents overfitting by using 80% data per tree
    colsample_bytree=0.8,                    # Uses 80% features per tree
    scale_pos_weight=scale_pos_weight_value, # Crucial for 85/15 class imbalance
    eval_metric="logloss"
)

# Train model
xgb_model.fit(X_train, y_train)

# Predict
y_pred_xgb = xgb_model.predict(X_test)

# Evaluate
xgb_results = evaluate_model(
    "XGBoost",
    y_test,
    y_pred_xgb
)
```

```python
# Build Neural Network

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Create Neural Network
nn_model = Sequential()

# Input Layer
nn_model.add(Dense(32, activation="relu", input_shape=(X_train_scaled.shape[1],)))

# Hidden Layer
nn_model.add(Dense(16, activation="relu"))

# Output Layer
nn_model.add(Dense(1, activation="sigmoid"))

# Compile the model
nn_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Display model summary
nn_model.summary()
```

```python
# Train Neural Network

history = nn_model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.2,
    epochs=20,
    batch_size=32,
    verbose=1
)
```

```python
#  Predict Using Neural Network

prediction = nn_model.predict(X_test_scaled)

# Convert probabilities into 0 and 1
y_pred_nn = (prediction > 0.5).astype(int)

nn_results = evaluate_model(
    "Neural Network",
    y_test,
    y_pred_nn
)
```

```python
# 1. Define sample student data using your dataset's original column names
good_student_data = {
    'Gender': 'Female',
    'Age': 20,
    'Study_Hours_Per_Week': 25,
    'Attendance_Percentage': 95.0,
    'CGPA': 3.8,
    'Stress_Level': 2,
    'Mental_Wellbeing': 8,
    'Family_Support': 'Yes',
    'Internet_Access': 'Yes',
    'Extracurricular': 'Yes',
    'Financial_Difficulty': 'No',
    'Screen_Time_Hours': 2,
    'Commute_Time_Minutes': 15,
    'Assignment_Completion': 98.0
}

bad_student_data = {
    'Gender': 'Male',
    'Age': 21,
    'Study_Hours_Per_Week': 5,
    'Attendance_Percentage': 50.0,
    'CGPA': 1.8,
    'Stress_Level': 9,
    'Mental_Wellbeing': 2,
    'Family_Support': 'No',
    'Internet_Access': 'No',
    'Extracurricular': 'No',
    'Financial_Difficulty': 'Yes',
    'Screen_Time_Hours': 8,
    'Commute_Time_Minutes': 60,
    'Assignment_Completion': 40.0
}

# 2. Function to align single inputs with X_train and predict
def predict_student_risk(student_dict, model, train_columns):
    # Convert dictionary to DataFrame
    df_raw = pd.DataFrame([student_dict])

    # One-Hot Encode categoricals
    df_encoded = pd.get_dummies(df_raw)

    # Reindex to match the exact training features and column order (fill missing dummy columns with 0)
    df_aligned = df_encoded.reindex(columns=train_columns, fill_value=0)

    # Predict probabilities
    proba = model.predict_proba(df_aligned)[0]
    return proba

# 3. Test both students
good_proba = predict_student_risk(good_student_data, xgb_model, X_train.columns)
bad_proba = predict_student_risk(bad_student_data, xgb_model, X_train.columns)

print(f"Good Student Probabilities [Class 0 (No Risk), Class 1 (At Risk)]:\n{good_proba}\n")
print(f"Bad Student Probabilities [Class 0 (No Risk), Class 1 (At Risk)]:\n{bad_proba}")
```
